**Dataset Description:**

The dataset contains groundwater level information collected from different states and districts in India.

 It includes details about groundwater depth, location, and changes in water levels over time.

The dataset helps analyze groundwater depletion, seasonal water level changes, and the impact of climate conditions on groundwater resources.

 It is useful for creating visualizations, identifying trends, and supporting groundwater conservation analysis.

**Type of Analysis**

**1. Descriptive Analysis**

Descriptive analysis summarizes the groundwater dataset using statistical measures such as average groundwater levels, count of wells, seasonal variations, and yearly changes. It helps in understanding the overall condition of groundwater resources across different states and districts. This provides a clear overview of groundwater patterns before performing deeper analysis.

**2. Diagnostic Analysis**

Diagnostic analysis focuses on identifying relationships and the reasons behind groundwater level changes. It uses comparisons, trends, and correlation analysis to answer “why” groundwater levels are increasing or decreasing in certain regions. For example, it explains how factors such as rainfall, agriculture usage, urbanization, and over-extraction affect groundwater depletion.

**3. Predictive Analysis**

Predictive analysis uses historical groundwater data to identify trends and forecast future groundwater conditions. It applies simple trend analysis and forecasting techniques to estimate whether groundwater levels may rise or decline in upcoming years. This helps in understanding future water availability and possible risk areas.

**4. Prescriptive Analysis**

Prescriptive analysis provides recommendations based on the insights obtained from the groundwater data. It suggests actions such as rainwater harvesting, controlled groundwater extraction, efficient irrigation systems, and water conservation policies. These recommendations help government authorities and communities make better decisions for sustainable groundwater management.

```python
print("=" * 50)
print("EXTRACTING DATE PARTS")
print("=" * 50)

# Make sure date is datetime
df['date'] = pd.to_datetime(df['date'], dayfirst=True)

# Extract all date parts
df['year']        = df['date'].dt.year
df['month']       = df['date'].dt.month
df['month_name']  = df['date'].dt.strftime('%B')
df['quarter']     = df['date'].dt.quarter
df['day']         = df['date'].dt.day
df['day_of_week'] = df['date'].dt.dayofweek
df['day_name']    = df['date'].dt.strftime('%A')
df['week_number'] = df['date'].dt.isocalendar().week.astype(int)
df['is_weekend']  = df['day_of_week'].apply(lambda x: 1 if x >= 5 else 0)

print("✅ Date parts extracted:")
print(df[['date','year','month','month_name',
          'quarter','day','day_name','is_weekend']].head())
```

```python
def get_season(month):
    if month in [12, 1, 2]:
        return 'Winter'
    elif month in [3, 4, 5]:
        return 'Summer'
    elif month in [6, 7, 8, 9]:
        return 'Monsoon'
    else:
        return 'Post-Monsoon'

df['season'] = df['month'].apply(get_season)

print("✅ Season column created:")
print(df['season'].value_counts())
```

```python
print("=" * 50)
print("WATER LEVEL DERIVED FEATURES")
print("=" * 50)

# Water status based on level_diff
df['water_status'] = df['level_diff'].apply(
    lambda x: 'Rising'  if x > 0 else
              'Falling' if x < 0 else 'Stable'
)

# Absolute change
df['abs_level_diff'] = df['level_diff'].abs()

# Level category based on depth
def categorize_level(level):
    if level <= 5:
        return 'Shallow'
    elif level <= 15:
        return 'Moderate'
    elif level <= 30:
        return 'Deep'
    else:
        return 'Very Deep'

df['level_category'] = df['currentlevel'].apply(categorize_level)

# Risk flag — falling AND deep
df['risk_flag'] = ((df['water_status'] == 'Falling') &
                   (df['level_category'].isin(['Deep','Very Deep']))
                  ).astype(int)

print("✅ Water level features created:")
print(df[['currentlevel','level_diff','water_status',
          'abs_level_diff','level_category','risk_flag']].head())
```

```python


print("=" * 50)
print(" GEOGRAPHIC DERIVED FEATURES")
print("=" * 50)

# Region based on latitude
def get_region(lat):
    if lat >= 28:
        return 'North India'
    elif lat >= 20:
        return 'Central India'
    elif lat >= 13:
        return 'South India'
    else:
        return 'Far South / Islands'

df['region'] = df['latitude'].apply(get_region)

# Distance from center of India (approx 22°N, 80°E)
df['dist_from_center'] = np.sqrt(
    (df['latitude']  - 22) ** 2 +
    (df['longitude'] - 80) ** 2
).round(2)

print("✅ Geographic features created:")
print(df[['latitude','longitude','region',
          'dist_from_center']].head())
```

```python


print("=" * 50)
print("STATISTICAL DERIVED FEATURES")
print("=" * 50)

# Z-score of currentlevel (detect outliers)
df['level_zscore'] = (
    (df['currentlevel'] - df['currentlevel'].mean()) /
     df['currentlevel'].std()
).round(3)

# Flag outliers (z-score > 3)
df['is_outlier'] = (df['level_zscore'].abs() > 3).astype(int)

# Binning currentlevel into 5 equal bins
df['level_bin'] = pd.cut(
    df['currentlevel'],
    bins=5,
    labels=['Very Low','Low','Medium','High','Very High']
)

print("✅ Statistical features created:")
print(df[['currentlevel','level_zscore',
          'is_outlier','level_bin']].head(8))
print(f"\nOutliers found: {df['is_outlier'].sum()}")
```

```python


fig, ax = plt.subplots(figsize=(10, 6))
for status, color in {'Rising':'#2ecc71',
                      'Falling':'#e74c3c',
                      'Stable':'#95a5a6'}.items():
    mask = df['water_status'] == status
    ax.scatter(df.loc[mask,'latitude'],
               df.loc[mask,'currentlevel'],
               c=color, label=status,
               alpha=0.6, s=50,
               edgecolors='white')
z = np.polyfit(df['latitude'], df['currentlevel'], 1)
p = np.poly1d(z)
x_line = np.linspace(df['latitude'].min(),
                     df['latitude'].max(), 100)
ax.plot(x_line, p(x_line),
        'k--', linewidth=2, label='Trend')
ax.set_title('Latitude vs Current Water Level',
             fontsize=14, fontweight='bold')
ax.set_xlabel('Latitude')
ax.set_ylabel('Current Level (m)')
ax.legend()
plt.tight_layout()
plt.show()

# ── INTERPRETATION ──────────────────────────────────────────
corr_val = df['latitude'].corr(df['currentlevel'])
strength = ('Strong' if abs(corr_val) > 0.5 else
            'Moderate' if abs(corr_val) > 0.3 else
            'Weak')
direction = 'positive' if corr_val > 0 else 'negative'

print("=" * 55)
print(" INTERPRETATION — Latitude vs Water Level")
print("=" * 55)
print(f"  → Correlation : {corr_val:.4f}")
print(f"  → Strength    : {strength} {direction} correlation")
if corr_val > 0:
    print("  → As latitude increases (moving north),")
    print("     groundwater depth also increases.")
    print("  → Northern India has deeper groundwater than south.")
else:
    print("  → As latitude increases (moving north),")
    print("     groundwater depth decreases.")
print(f"  → Insight: {strength} geographic pattern detected.")
print("     Policy makers should target high-latitude")
print("     regions for groundwater conservation programs.")
```
